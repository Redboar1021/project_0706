```python
import numpy as np
import plotly.graph_objects as go

# 点群生成
points = np.random.rand(1, 6890, 3)[0]

# 3Dプロット
fig = go.Figure(data=[go.Scatter3d(
    x=points[:, 0], y=points[:, 1], z=points[:, 2],
    mode='markers',
    marker=dict(size=2, color='blue')
)])
fig.update_layout(title='3D Point Cloud')
fig.show()
```
